```python
import torch, numpy as np

sd = torch.load("models/PPO_policy_0.744.pth", map_location="cpu")

def emit(name, arr):
    arr = arr.detach().cpu().numpy().astype(np.float32)
    flat = ", ".join(f"{x:.8f}f" for x in arr.flatten())
    shape = "][".join(str(d) for d in arr.shape)
    print(f"const float {name}[{arr.size}] = {{{flat}}}; // shape {arr.shape}")

for k, v in sd.items():
    emit(k.replace(".", "_"), v)
```
